```python
input_sdf = "generated_decoys_activeER.sdf"
output_sdf = "generated_decoys_activeER_filtered_FINAL.sdf"

remove_idx = {107, 163, 189}  # use a set for multiple indices

ligands = []
current = []

with open(input_sdf) as f:
    for line in f:
        current.append(line)
        if line.strip() == "$$$$":
            ligands.append("".join(current))
            current = []

# remove both ligands
filtered = [lig for i, lig in enumerate(ligands) if i not in remove_idx]

# write new SDF
with open(output_sdf, "w") as f:
    for lig in filtered:
        f.write(lig)

print("✅ Done. Total ligands:", len(filtered))
```
