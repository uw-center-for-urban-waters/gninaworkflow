```python
from rdkit import Chem
from rdkit.Chem import rdMolAlign, inchi
import sys

def comparesdf(f1, f2):
    #f1, f2 = sys.argv[1], sys.argv[2]
    #f1 = "481-30-1.sdf"
    #f2 = "cas-481-30-1_min.sdf"
    
    m1 = Chem.MolFromMolFile(f1, removeHs=False, sanitize=True)
    m2 = Chem.MolFromMolFile(f2, removeHs=False, sanitize=True)
    
    if m1 is None or m2 is None:
        raise SystemExit("Failed to read one or both SDF files")
    
    s1 = Chem.MolToSmiles(m1, canonical=True, isomericSmiles=True)
    s2 = Chem.MolToSmiles(m2, canonical=True, isomericSmiles=True)
    
    same_smiles = s1 == s2
    same_inchi = inchi.MolToInchi(m1) == inchi.MolToInchi(m2)
    
    rmsd = None
    if same_smiles and m1.GetNumAtoms() == m2.GetNumAtoms():
        try:
            rmsd = rdMolAlign.AlignMol(m2, m1)
        except Exception:
            rmsd = None
    
    print("mol1_smiles:", s1)
    print("mol2_smiles:", s2)
    print("same_canonical_smiles:", same_smiles)
    print("same_inchi:", same_inchi)
    print("num_atoms_mol1:", m1.GetNumAtoms())
    print("num_atoms_mol2:", m2.GetNumAtoms())
    print("aligned_rmsd:", rmsd)
```

```python
from openbabel import openbabel as ob
from rdkit import Chem

def writeonedecoytoFS(row):
    
    # Your RDKit mol
    rdkit_mol = row['ROMol']

    # Convert RDKit MolBlock → Open Babel OBMol
    molblock = Chem.MolToMolBlock(rdkit_mol)
    ob_mol = ob.OBMol()
    ob_conversion = ob.OBConversion()
    ob_conversion.SetInFormat("mdl")  # MOL block format
    ob_conversion.ReadString(ob_mol, molblock)

    # Write out GNINA-ready SDF
    ob_conversion.SetOutFormat("sdf")
    sdf_string = ob_conversion.WriteString(ob_mol)

    with open(decoydir + "decoyligand.sdf", "w") as f:
        f.write(sdf_string)
```
