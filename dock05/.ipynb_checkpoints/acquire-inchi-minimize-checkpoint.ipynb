```python
# Use InChiKey to download meta data from PubChem that includes a unique CID identifier.
import pandas as pd
import pubchempy as pcp

outdf = None

def ask_pubchem(inchifile):
    # Input CSV must have a column named 'InChiKey'
    df = pd.read_csv(inchifile)
    results = []
    #print(df) 
    for ikey in df["InChIKey"].dropna():
        ikey = str(ikey).strip()    
        try:
            #compounds = pcp.get_compounds(cas, namespace="name")
            compounds = pcp.get_compounds(ikey, 'inchikey')
            if compounds:
                c = compounds[0]
                results.append({
                    "CID": c.cid,
                    "InChIKey": c.inchikey,
                    "MolecularFormula": c.molecular_formula,
                    "MolecularWeight": c.molecular_weight,
                    "IUPACName": c.iupac_name
                })
            else:
                results.append({
                    "CID": None,
                    "InChIKey": None,
                    "MolecularFormula": None,
                    "MolecularWeight": None,
                    "IUPACName": None
                })
        except Exception as e:
            results.append({
                "CID": None,
                "InChIKey": None,
                "MolecularFormula": None,
                "MolecularWeight": None,
                "IUPACName": f"ERROR: {e}"
            })

    return (results)
```

```python
import pubchempy as pcp


def lookup_by_inchikey(inchikey: str):
    compounds = pcp.get_compounds(inchikey, 'inchikey')
    return compounds[0] if compounds else None


if __name__ == '__main__':
    key = 'BSYNRYMUTXBXSQ-UHFFFAOYSA-N'  # aspirin
    compound = lookup_by_inchikey(key)

    if compound:
        print('CID:', compound.cid)
        print('Name:', compound.iupac_name)
        print('MolecularFormula:', compound.molecular_formula)
        print('MolecularWeight:', compound.molecular_weight)
        #print('SMILES:', compound.canonical_smiles)
        print('SMILES:', compound.connectivity_smiles)
    else:
        print('No compound found for that InChI Key.')
```

```python
# Create and then append a list with metadata based on lists of InChiKey.
pubchemresults = []
pubchemresults.extend(ask_pubchem(inchikeyfile))
print(pubchemresults)
```

```python
import csv

# Path to your CSV file
csv_file = pubchemdir + "pubchem_lookup_results.csv"

# Read single-column CSV into a Python list
cas_list = []

with open(csv_file, newline='', encoding='utf-8') as f:
    reader = csv.reader(f)

    for row in reader:
        if row:                 # skip empty rows
            cas_list.append(row[0])

print(cas_list)
```

```python
import time

def fetch_with_retry(cid, max_tries=5, base_delay=2):
    for attempt in range(max_tries):
        try:
            #return get_sdf(cid)
            return pcp.download("SDF", out, cid, namespace='cid', overwrite=True, record_type='3d')
        except Exception as e:
            msg = str(e)
            if "502" not in msg:
                raise
            if attempt == max_tries - 1:
                raise
            time.sleep(base_delay * (2 ** attempt))
```

```python
# Let's download the 3d versions of .sdf files from PubChem.
import pubchempy as pcp
import time
from pathlib import Path

if cas_list and cas_list[0] == 'CAS':
    cas_list.pop(0)
    
#print(cas_list)
#cas_list = ["120-47-8","72-55-9","72-43-5"]

for cas in cas_list:
    outsdf = pubchemdir + sdffiles + "cas-" + cas + ".sdf"
    outsdf_path = Path(outsdf)
    if outsdf_path.is_file():
        print("File " + outsdf + " already present.")
    else:
        matches = pcp.get_compounds(cas, namespace="name")
        if matches:
            cid = matches[0].cid
            out = pubchemdir + sdffiles + "cas-" + cas + ".sdf"
            print ("Download and write " + out)
            pcp.download("SDF", outsdf, cid, namespace='cid', overwrite=True, record_type='3d')
            time.sleep(2.0)
            #fetch_with_retry(cid)
        else:
            print("No match found for " + cas + ". Sorry. :(")
       
```
