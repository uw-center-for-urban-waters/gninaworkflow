## Lookup and Download Pubchem files by CAS number

```python
# Use CAS numbers to download meta data from PubChem that includes a unique CID identifier.
import pandas as pd
import pubchempy as pcp

outdf = None

def ask_pubchem(casfile):
    # Input CSV must have a column named 'CAS'
    df = pd.read_csv(casfile)
    
    results = []
    
    for cas in df["CAS"].dropna():
        cas = str(cas).strip()
        try:
            compounds = pcp.get_compounds(cas, namespace="name")
            if compounds:
                c = compounds[0]
                results.append({
                    "CAS": cas,
                    "CID": c.cid,
                    "MolecularFormula": c.molecular_formula,
                    "MolecularWeight": c.molecular_weight,
                    "IUPACName": c.iupac_name
                })
            else:
                results.append({
                    "CAS": cas,
                    "CID": None,
                    "MolecularFormula": None,
                    "MolecularWeight": None,
                    "IUPACName": None
                })
        except Exception as e:
            results.append({
                "CAS": cas,
                "CID": None,
                "MolecularFormula": None,
                "MolecularWeight": None,
                "IUPACName": f"ERROR: {e}"
            })

    return (results)
```

```python
# Create and then append a list with metadata based on lists of CAS numbers.
pubchemresults = []
pubchemresults.extend(ask_pubchem(weakcasfile))
print(results)
```

```python
pubchemresults.extend(ask_pubchem(moderatecasfile))
print(pubchemresults)
```
