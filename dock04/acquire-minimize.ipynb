# Acquire and Minimize

## Configuration

```python
#inputeliganddir = "preprocessed/Ideal_Ligand/"
outputminimizeddir = "preprocessed/minimized_ideal_ligand/"
pubchemdir = "preprocessed/pubchem/"
casdir = "moderate/"
casfile = pubchemdir + "cas-moderate.csv"
```

## Lookup and Download Pubchem files by CAS number

```python
import pandas as pd
import pubchempy as pcp

# Input CSV must have a column named 'CAS'
df = pd.read_csv(casfile)

results = []

for cas in df["CAS"].dropna():
    cas = str(cas).strip()
    try:
        compounds = pcp.get_compounds(cas, namespace="name")
        if compounds:
            c = compounds[0]
            results.append({
                "CAS": cas,
                "CID": c.cid,
                "MolecularFormula": c.molecular_formula,
                "MolecularWeight": c.molecular_weight,
                "IUPACName": c.iupac_name
            })
        else:
            results.append({
                "CAS": cas,
                "CID": None,
                "MolecularFormula": None,
                "MolecularWeight": None,
                "IUPACName": None
            })
    except Exception as e:
        results.append({
            "CAS": cas,
            "CID": None,
            "MolecularFormula": None,
            "MolecularWeight": None,
            "IUPACName": f"ERROR: {e}"
        })

out = pd.DataFrame(results)
out.to_csv(pubchemdir + "pubchem_lookup_results.csv", index=False)
```

```python
import csv

# Path to your CSV file
csv_file = pubchemdir + "pubchem_lookup_results.csv"

# Read single-column CSV into a Python list
cas_list = []

with open(csv_file, newline='', encoding='utf-8') as f:
    reader = csv.reader(f)

    for row in reader:
        if row:                 # skip empty rows
            cas_list.append(row[0])

print(cas_list)
```

```python
import pubchempy as pcp

if cas_list and cas_list[0] == 'CAS':
    cas_list.pop(0)
print(cas_list)

#cas_list = ["120-47-8","72-55-9","72-43-5"]

for cas in cas_list:
    matches = pcp.get_compounds(cas, namespace="name")
    if matches:
        cid = matches[0].cid
        out = pubchemdir + casdir + "cas-" + cas + ".sdf"
        print ("Download and write " + out)
        pcp.download("SDF", out, cid, namespace='cid', overwrite=True, record_type='3d')
    else:
        print("No match found for " + cas + ". Sorry. :(")
```
